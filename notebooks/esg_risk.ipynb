```python
def fetch_company_filings(ticker: str, forms=("10-K", "10-Q"), limit=5):
    company = Company(ticker)

    rows= []
    for form in forms:
        filings= company.get_filings(form=form).head(limit)

        for filing in filings:
            rows.append({
                "ticker": ticker,
                "company": filing.company,
                "form_type": form,
                "filing_date": filing.filing_date,
                "accession_number": filing.accession_number,
                "text": filing.text()
            })

        return pd.DataFrame(rows)
        
        
    


```

```python
sections = {}

for item in tenk.items:
    try:
        section_text = tenk[item]
        sections[item] = section_text
        print(item, ":", len(section_text.split()), "words")
    except Exception as e:
        print(item, "failed:", e)
```

```python
import pandas as pd

section_df = pd.DataFrame([
    {
        "item": item,
        "text": text,
        "word_count": len(text.split())
    }
    for item, text in sections.items()
])

section_df
```

```python
import re

def clean_filing_text(text: str) -> str:
    if not isinstance(text, str):
        return ""

    # Replace non-breaking spaces
    text = text.replace("\xa0", " ")

    # Normalize smart quotes lightly
    text = text.replace("’", "'").replace("“", '"').replace("”", '"')

    # Add missing space between lowercase/period and uppercase words
    # Example: "DisclosuresNot applicable" -> "Disclosures Not applicable"
    text = re.sub(r"([a-z])([A-Z])", r"\1 \2", text)

    # Add space between period and next capital letter if missing
    # Example: "applicable.Item" -> "applicable. Item"
    text = re.sub(r"\.([A-Z])", r". \1", text)

    # Collapse repeated whitespace but keep text readable
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()
```

```python
section_df["clean_text"] = section_df["text"].apply(clean_filing_text)
section_df["clean_word_count"] = section_df["clean_text"].apply(lambda x: len(x.split()))
```
